```python
import csv
import numpy as np
with open('csv/trainMapsGreedy/run-lightning_logs_trainMapsGreedy-tag-metrics_p_exp.csv', mode ='r')as file:
  csvFile = csv.reader(file)
  data = []
  i = 0
  for lines in csvFile:
    if i > 0:
       value = float(lines[2])
       data.insert(0, value)
       print(data)
    i += 1
        # Parametri del bootstrap
  print(data)
  n_iterations = 100
  n_size = len(data)

    # Distribuzione bootstrap
  bootstrap_samples = [np.mean(np.random.choice(data, size=n_size, replace=True)) for _ in range(n_iterations)]
  bootstrap_mean = np.mean(bootstrap_samples)

    # Limiti di confidenza al 95%
  lower_bound = np.percentile(bootstrap_samples, 2.5)
  upper_bound = np.percentile(bootstrap_samples, 97.5)

  print("Bootstrap Mean:", bootstrap_mean)
  print("95% Confidence Interval:", (lower_bound, upper_bound))

   
```
